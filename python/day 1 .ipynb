```python
class SmartPhone:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        self.__battery = 100 

    def use_app(self, minutes):
        drain = minutes * 0.5
        self.__battery = max(0, self.__battery - drain)

    def charge(self):
        self.__battery = 100

    def status(self):
        print(f"Brand: {self.brand} | Model: {self.model} | Battery: {self.__battery}%")


class GamingPhone(SmartPhone):
    def __init__(self, brand, model, cooling_mode=False):
        super().__init__(brand, model)
        self.cooling_mode = cooling_mode

    def boost_performance(self):
        self.cooling_mode = True
        print(f"{self.brand} {self.model}: Performance boosted! Cooling mode enabled.")

phone = SmartPhone("Samsung", "Galaxy S24")
phone.use_app(50)        
phone.status()
phone.charge()         
phone.status()
```
